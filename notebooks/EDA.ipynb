```python
import pandas as pd
import plotly_express as px

# Carregando os dados (voltando uma pasta para achar o CSV na raiz)
df = pd.read_csv('../vehicles_us.csv')

# Exibir as primeiras linhas e informações gerais sobre o DataFrame
display(df.head())
df.info()
```

```python
# Criar e exibir o histograma usando a variável 'df' (sem ler o arquivo de novo)
fig = px.histogram(df, x="odometer")
fig.show()
```

```python
# Criar e exibir o gráfico de dispersão usando a mesma variável 'df'
fig = px.scatter(df, x="odometer", y="price")
fig.show()
```
